```python
import pandas as pd
import numpy as np

df = pd.read_csv("creditcard_2023.csv")

print("Dataset Shape:", df.shape)

print("\nFirst 5 Records:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# ==========================================
# PHASE 2: TARGET VARIABLE ANALYSIS
# ==========================================

# Count transaction classes
class_counts = df['Class'].value_counts()

print("Transaction Class Distribution:")
print(class_counts)

# Calculate percentages
class_percentages = df['Class'].value_counts(normalize=True) * 100

print("\nClass Distribution (%):")
print(class_percentages.round(2))

# Display fraud and legitimate transactions
print("\nLegitimate Transactions:", class_counts.get(0, 0))
print("Fraudulent Transactions:", class_counts.get(1, 0))

# Check unique values
print("\nUnique Class Values:")
print(df['Class'].unique())
```

```python
#visualizing fraud distribution

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Class')

plt.title('Credit Card Fraud vs Legitimate Transactions')
plt.xlabel('Transaction Class')
plt.ylabel('Number of Transactions')

plt.xticks([0, 1], ['Legitimate', 'Fraudulent'])

plt.show()
```

```python
# Statistical analysis of transaction amounts

print("Transaction Amount Statistics:")
display(df['Amount'].describe())

# Fraud vs legitimate transaction amounts
print("\nTransaction Amount by Class:")
display(df.groupby('Class')['Amount'].describe())
```

```python
# ==========================================
# PHASE 3: FEATURE QUALITY INVESTIGATION
# ==========================================

# Check whether transaction IDs are unique
print("Unique Transaction IDs:", df['id'].nunique())

# Check correlation between ID and fraud class
id_correlation = df['id'].corr(df['Class'])

print("\nID and Class Correlation:")
print(round(id_correlation, 4))

# Examine ID ranges for each class
print("\nID Distribution by Transaction Class:")

display(
    df.groupby('Class')['id'].agg(
        ['min', 'max', 'mean', 'median']
    )
)

# Feature statistical summary
print("\nFeature Statistical Summary:")

display(df.drop(columns=['id', 'Class']).describe().T)
```

```python
# ==========================================
# FEATURE CORRELATION ANALYSIS
# ==========================================

correlation = df.drop(columns=['id']).corr()

fraud_correlation = correlation['Class'].drop('Class')

fraud_correlation = fraud_correlation.sort_values(
    ascending=False
)

print("Features Most Positively Associated with Fraud:")
display(fraud_correlation.head(10))

print("\nFeatures Most Negatively Associated with Fraud:")
display(fraud_correlation.tail(10))
```

```python
#visualizing the results


plt.figure(figsize=(12, 8))

fraud_correlation.sort_values().plot(
    kind='barh'
)

plt.title('Feature Correlation with Credit Card Fraud')
plt.xlabel('Correlation Coefficient')
plt.ylabel('Features')

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# TRANSACTION ID INVESTIGATION
# ==========================================

print("Unique Transaction IDs:")
print(df['id'].nunique())

print("\nID Correlation with Fraud Class:")
print(round(df['id'].corr(df['Class']), 4))

print("\nID Statistics by Transaction Class:")

display(
    df.groupby('Class')['id'].agg(
        ['min', 'max', 'mean', 'median']
    )
)
```

```python
# ==========================================
# PHASE 4: DATA PREPROCESSING
# ==========================================

# Separate independent and dependent variables

X = df.drop(columns=['id', 'Class'])

y = df['Class']

# Check dimensions

print("Feature Matrix Shape:", X.shape)
print("Target Variable Shape:", y.shape)

# Verify that ID has been removed

print("\nRemaining Features:")
print(X.columns.tolist())

# Confirm target distribution

print("\nTarget Distribution:")
print(y.value_counts())
```

```python
# ==========================================
# TRAIN-TEST SPLIT
# ==========================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("\nTraining Target:", y_train.shape)
print("Testing Target:", y_test.shape)

print("\nTraining Class Distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting Class Distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
# ==========================================
# PHASE 6: FEATURE SCALING
# ==========================================

from sklearn.preprocessing import StandardScaler

# Initialise scaler
scaler = StandardScaler()

# Fit and transform training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform testing data
X_test_scaled = scaler.transform(X_test)

# Verify scaled data
print("Training Data Shape:", X_train_scaled.shape)
print("Testing Data Shape:", X_test_scaled.shape)

print("\nTraining Feature Mean:")
print(X_train_scaled.mean(axis=0)[:5].round(4))

print("\nTraining Feature Standard Deviation:")
print(X_train_scaled.std(axis=0)[:5].round(4))
```

```python
from sklearn.linear_model import LogisticRegression

# Initialise Logistic Regression
log_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

print("Logistic Regression model initialised!")
```

```python
# Train Logistic Regression

log_model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully!")
```

```python
y_pred_log = log_model.predict(X_test_scaled)

print("Predictions generated successfully!")
print(y_pred_log[:20])
```

```python
# ==========================================
# MODEL EVALUATION
# ==========================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)

# Calculate performance metrics

accuracy = accuracy_score(y_test, y_pred_log)
precision = precision_score(y_test, y_pred_log)
recall = recall_score(y_test, y_pred_log)
f1 = f1_score(y_test, y_pred_log)

y_prob_log = log_model.predict_proba(X_test_scaled)[:, 1]

roc_auc = roc_auc_score(y_test, y_prob_log)

print("LOGISTIC REGRESSION PERFORMANCE")
print("--------------------------------")

print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1 Score:  {f1:.4f}")
print(f"ROC-AUC:   {roc_auc:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred_log))
```

```python
# ==========================================
# PHASE 8: CONFUSION MATRIX
# ==========================================

from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

cm = confusion_matrix(y_test, y_pred_log)

print("Confusion Matrix:")
print(cm)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Legitimate', 'Fraudulent'],
    yticklabels=['Legitimate', 'Fraudulent']
)

plt.title('Logistic Regression - Confusion Matrix')
plt.xlabel('Predicted Class')
plt.ylabel('Actual Class')

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# PHASE 9: RANDOM FOREST
# ==========================================

from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=None,
    min_samples_split=2,
    random_state=42,
    n_jobs=-1
)

print("Random Forest model initialised!")
```

```python
# ==========================================
# RANDOM FOREST EVALUATION
# ==========================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)

print("RANDOM FOREST PERFORMANCE")
print("--------------------------------")

print(f"Accuracy:  {accuracy_score(y_test, y_pred_rf):.4f}")
print(f"Precision: {precision_score(y_test, y_pred_rf):.4f}")
print(f"Recall:    {recall_score(y_test, y_pred_rf):.4f}")
print(f"F1 Score:  {f1_score(y_test, y_pred_rf):.4f}")
print(f"ROC-AUC:   {roc_auc_score(y_test, y_prob_rf):.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
# ==========================================
# DATA QUALITY: DUPLICATE FEATURE CHECK
# ==========================================

# Remove transaction ID because it is not a predictive feature
audit_df = df.drop(columns=['id'])

# Check duplicate feature and target combinations
duplicate_rows = audit_df.duplicated().sum()

# Check duplicate feature combinations regardless of target
duplicate_features = audit_df.drop(
    columns=['Class']
).duplicated().sum()

print("DATA DUPLICATE INVESTIGATION")
print("--------------------------------")

print("Total duplicate records:", duplicate_rows)

print("Duplicate feature combinations:", duplicate_features)

# Check whether identical features have conflicting labels
conflicting_labels = (
    audit_df.groupby(list(audit_df.columns[:-1]))['Class']
    .nunique()
    .gt(1)
    .sum()
)

print("Conflicting feature patterns:", conflicting_labels)
```

```python
# ==========================================
# TRANSACTION ID DISTRIBUTION AUDIT
# ==========================================

# Divide transaction IDs into four groups
df['ID_Group'] = pd.qcut(
    df['id'],
    q=4,
    labels=['Q1', 'Q2', 'Q3', 'Q4']
)

# Examine fraud distribution across ID groups
id_distribution = pd.crosstab(
    df['ID_Group'],
    df['Class'],
    normalize='index'
) * 100

id_distribution.columns = [
    'Legitimate (%)',
    'Fraudulent (%)'
]

print("FRAUD DISTRIBUTION BY TRANSACTION ID RANGE")

display(id_distribution.round(2))
```

```python
# ==========================================
# TRANSACTION ID AND FRAUD RELATIONSHIP
# ==========================================

# Calculate average transaction ID by class

id_analysis = df.groupby('Class')['id'].agg(
    ['count', 'min', 'max', 'mean', 'median']
)

print("TRANSACTION ID STATISTICS BY CLASS")

display(id_analysis.round(2))

# Visualise transaction ID distribution

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='Class',
    y='id'
)

plt.title('Transaction ID Distribution by Fraud Class')
plt.xlabel('Transaction Class (0 = Legitimate, 1 = Fraud)')
plt.ylabel('Transaction ID')

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# PHASE 11: DATASET ORDERING INVESTIGATION
# ==========================================

# Sort records by transaction ID
sorted_df = df.sort_values('id').copy()

# Check class transitions in ID order
class_changes = (
    sorted_df['Class'] != sorted_df['Class'].shift()
).sum() - 1

print("DATASET ORDERING INVESTIGATION")
print("--------------------------------")

print("First 20 transaction classes:")
print(sorted_df['Class'].head(20).tolist())

print("\nLast 20 transaction classes:")
print(sorted_df['Class'].tail(20).tolist())

print("\nNumber of class transitions:")
print(class_changes)

# Investigate fraud percentage across 10 ID ranges
sorted_df['ID_Range'] = pd.qcut(
    sorted_df['id'],
    q=10,
    labels=False
)

id_class_analysis = pd.crosstab(
    sorted_df['ID_Range'],
    sorted_df['Class'],
    normalize='index'
) * 100

id_class_analysis.columns = [
    'Legitimate (%)',
    'Fraudulent (%)'
]

print("\nFRAUD DISTRIBUTION ACROSS 10 ID RANGES")

display(id_class_analysis.round(2))
```

```python
# ==========================================
# PHASE 12: CROSS-VALIDATION
# ==========================================

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.ensemble import RandomForestClassifier

print("Cross-validation libraries imported successfully!")
```

```python
# Separate features and target

X_cv = df.drop(
    columns=['id', 'Class', 'ID_Group'],
    errors='ignore'
)

y_cv = df['Class']

print("Feature Matrix:", X_cv.shape)
print("Target Variable:", y_cv.shape)

print("\nTarget Distribution:")
print(y_cv.value_counts())
```

```python
# Initialise Random Forest

rf_cv_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=1
)

# Configure Stratified 5-Fold Cross-Validation

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("Random Forest model configured successfully!")
print("Cross-validation folds: 5")
```

```python
# Perform cross-validation

cv_results = cross_validate(
    rf_cv_model,
    X_cv,
    y_cv,
    cv=cv,
    scoring=[
        'accuracy',
        'precision',
        'recall',
        'f1',
        'roc_auc'
    ],
    n_jobs=-1
)

print("Cross-validation completed successfully!")
```

```python
# Calculate average performance

metrics = [
    'accuracy',
    'precision',
    'recall',
    'f1',
    'roc_auc'
]

print("RANDOM FOREST: 5-FOLD CROSS-VALIDATION")
print("======================================")

for metric in metrics:

    scores = cv_results[f'test_{metric}']

    print(
        f"{metric.upper()}: "
        f"{scores.mean():.4f} "
        f"+/- {scores.std():.4f}"
    )
```

```python
# ==========================================
# PHASE 13: INDIVIDUAL FOLD PERFORMANCE
# ==========================================

import pandas as pd

fold_results = pd.DataFrame({
    'Fold': [1, 2, 3, 4, 5],
    'Accuracy': cv_results['test_accuracy'],
    'Precision': cv_results['test_precision'],
    'Recall': cv_results['test_recall'],
    'F1 Score': cv_results['test_f1'],
    'ROC-AUC': cv_results['test_roc_auc']
})

print("INDIVIDUAL CROSS-VALIDATION RESULTS")

display(fold_results.round(5))

print("\nAVERAGE PERFORMANCE")

display(
    fold_results.drop(columns='Fold')
    .mean()
    .round(5)
)
```

```python
# ==========================================
# PHASE 13: MODEL ROBUSTNESS TESTING
# TRAINING VS TESTING PERFORMANCE
# ==========================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
import pandas as pd

# Logistic Regression - Training predictions
y_train_pred_log = log_model.predict(X_train_scaled)
y_train_prob_log = log_model.predict_proba(X_train_scaled)[:, 1]

# Random Forest - Training predictions
y_train_pred_rf = rf_model.predict(X_train)
y_train_prob_rf = rf_model.predict_proba(X_train)[:, 1]

# Create performance comparison
robustness_results = []

models = {
    "Logistic Regression": {
        "train_pred": y_train_pred_log,
        "test_pred": y_pred_log,
        "train_prob": y_train_prob_log,
        "test_prob": y_prob_log
    },
    "Random Forest": {
        "train_pred": y_train_pred_rf,
        "test_pred": y_pred_rf,
        "train_prob": y_train_prob_rf,
        "test_prob": y_prob_rf
    }
}

for model_name, results in models.items():

    for dataset_name, actual, predicted, probability in [
        ("Training", y_train, results["train_pred"],
         results["train_prob"]),

        ("Testing", y_test, results["test_pred"],
         results["test_prob"])
    ]:

        robustness_results.append({
            "Model": model_name,
            "Dataset": dataset_name,
            "Accuracy": accuracy_score(actual, predicted),
            "Precision": precision_score(actual, predicted),
            "Recall": recall_score(actual, predicted),
            "F1 Score": f1_score(actual, predicted),
            "ROC-AUC": roc_auc_score(actual, probability)
        })

# Display results
robustness_df = pd.DataFrame(robustness_results)

print("MODEL ROBUSTNESS: TRAINING VS TESTING")
print("======================================")

display(robustness_df.round(4))
```
